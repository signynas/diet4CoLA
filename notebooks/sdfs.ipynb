# Signed Distance Functions

This notebook highlights some of the **signed distance functions** (SDFs) we might want to use for generating synthetic CoLA cuts.

```python
from typing import Callable

import numpy as np
```

Make sure to run `pip install -e .` before running the cell below. This command will make sure `diet4cola` is installed as a module in your current Python/Conda environment.

```python
from diet4cola.sdf import sdf_segment, sdf_box, sdf_capsule, sdf_vesica
from diet4cola.utils import plot_sdf
```

### Utility Functions

Here we define some utility functions for this notebook.

```python
def compute_sdf(width: int,
                height: int,
                origin: tuple[int, int],
                destination: tuple[int, int],
                parameter: float,
                sdf: Callable) -> np.ndarray:
    A       = np.array([origin[0], origin[1]])
    B       = np.array([destination[0], destination[1]])
    data    = np.zeros([height, width])
    
    for y in range(height):
        for x in range(width):
            P = np.array([x, y])
            data[y, x] = sdf(P, A, B, parameter)

    return data
```

```python
origin = (120, 370)
destination = (370, 120)
```

### Segment SDF

```python
segment = compute_sdf(512, 512, origin, destination, 0, sdf_segment)
plot_sdf(segment, origin, destination, 'Segment SDF', cmap='RdBu_r')
```

### Box SDF

```python
box = compute_sdf(512, 512, origin, destination, 150, sdf_box)
plot_sdf(box, origin, destination, 'Box SDF', cmap='RdBu_r')
```

### Capsule SDF

```python
capsule = compute_sdf(512, 512, origin, destination, 50, sdf_capsule)
plot_sdf(capsule, origin, destination, 'Capsule SDF', cmap='RdBu_r')
```

### Vesica SDF

```python
vesica = compute_sdf(512, 512, origin, destination, 50, sdf_vesica)
plot_sdf(vesica, origin, destination, 'Vesica SDF', cmap='RdBu_r')
```

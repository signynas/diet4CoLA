```python
from diet4cola.utils import load_array, plot_2d_array
from pathlib import Path

import numpy as np
```

```python
KEY = 'cortex_SCci5ZCwrGMDPEEEpwizmx'
KEY_INPUT_STRING = f'{KEY}_actomyosin.npy'
KEY_OUTPUT_STRING = f'{KEY}_velocity_field.npy'
KEY_OUTPUT_STRING_DX = f'{KEY}_velocity_field_dx.npy'
KEY_OUTPUT_STRING_DY = f'{KEY}_velocity_field_dy.npy'
KEY_BOUNDARY_STRING = f'{KEY}_boundary.npy'

data_dir = Path('../data/synthetic')
```

```python
frame = 6
```

```python
input_example_file = data_dir / KEY_INPUT_STRING
output_example_file = data_dir / KEY_OUTPUT_STRING
output_example_file_dx = data_dir / KEY_OUTPUT_STRING_DX
output_example_file_dy = data_dir / KEY_OUTPUT_STRING_DY
boundary_file = data_dir / KEY_BOUNDARY_STRING

input_example = load_array(str(input_example_file))
output_example = load_array(str(output_example_file))
output_example_dx = load_array(str(output_example_file_dx))
output_example_dy = load_array(str(output_example_file_dy))
boundary = load_array(str(boundary_file))
```

```python
i_1 = input_example[frame, :, :]
i_2 = input_example[frame + 1, :, :]
vf = output_example[frame, :, :]
vf_dx = output_example_dx[frame, :, :]
vf_dy = output_example_dy[frame, :, :]
```

```python
plot_2d_array(i_1, 'Input i')
plot_2d_array(i_2, 'Input i+1')
```

```python
plot_2d_array(vf, 'Output i', 'viridis', np.min(vf), np.max(vf))
```

```python
plot_2d_array(vf_dx, 'Output i dX', 'viridis', np.min(vf_dx), np.max(vf_dx))
plot_2d_array(vf_dy, 'Output i dY', 'viridis', np.min(vf_dy), np.max(vf_dy))
```

```python
plot_2d_array(boundary, 'Cell Boundary')
```
